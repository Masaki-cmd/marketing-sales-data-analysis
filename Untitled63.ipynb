## データクレンジングと前処理

### Subtask:
データの整合性を保つため、欠損値、重複、誤ったデータ型、不整合な値などの「汚いデータ」を処理します。処理の過程で「データが汚かったので、このように処理して整合性を保った」という具体的な記述をテキストセルに追加します。


**Reasoning**:
The first step is to convert the 'date' column to a datetime object, check and drop duplicate rows, and then impute missing numerical values with their medians as per the instructions.



````python
import numpy as np

# 1. 'date' 列を datetime オブジェクトに変換
df['date'] = pd.to_datetime(df['date'])
print("Converted 'date' column to datetime.")

# 2. データセットに重複行がないか確認し、もしあれば削除
duplicate_rows_before = df.duplicated().sum()
if duplicate_rows_before > 0:
    df.drop_duplicates(inplace=True)
    print(f"Removed {duplicate_rows_before} duplicate rows.")
else:
    print("No duplicate rows found.")

# 3. 欠損値を適切に処理
# 数値型の欠損値は中央値で補完するのが一般的であり、外れ値に強いという利点があります。
# 特にパーセンテージやスコア、期間に関するデータなので、平均値よりも中央値が適切と判断します。
missing_cols = ['discount_percentage', 'customer_satisfaction_score', 'email_open_rate', 'days_since_last_purchase']
for col in missing_cols:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
        print(f"Filled missing values in '{col}' with median: {median_val:.2f}")

print("\nDataFrame info after cleaning:")
df.info()
print("\nDataFrame head after cleaning:")
print(df.head())
````

### 投資対効果（ROI）の評価：概要

本ステップでは、前段階で提案された各施策について、その**投資対効果（ROI）**を評価します。ROIの評価は、以下の観点に基づいて行われます。

1.  **施策の再確認**: 提案された具体的なアクションを再度確認します。
2.  **予測される売上インパクトの仮説設定**: 各施策が成功した場合に期待される売上増加額（またはコスト削減額）を仮定します。
3.  **予測される工数（コスト）の仮説設定**: 各施策の実施にかかる費用（人件費、ツール費用、広告費など）を仮定します。
4.  **ROIの算出**: 一般的なROIの計算式 `(売上増加額 - コスト) / コスト * 100%` を用いてROIを算出します。
5.  **ビジネス視点での実施の正当性（Justification）**: 算出したROIに基づき、その施策がビジネスにとってどれほど魅力的であるかを記述します。
6.  **優先順位付けの基準定義**: 複数の施策の中から、どの施策を優先して実行すべきかを判断するための基準を提案します。

**仮説設定の前提**:
今回の評価では、具体的な数値データがないため、各項目について現実的な範囲での**仮想的な数値**を用いて評価を行います。これらの数値はあくまで例示であり、実際のビジネスにおいては、より詳細な調査、市場データ、過去の実績、専門家の意見などを基に精緻な見積もりを行う必要があります。

それでは、各施策グループごとにROI評価を進めていきましょう。

#### 優先順位付けの基準定義

複数の魅力的な施策がある中で、限られたリソースを最も効果的に配分するためには、明確な優先順位付けが必要です。以下の基準を複合的に考慮し、施策の実施順序を決定することを推奨します。

1.  **ROIの高さ**: 最も高い投資対効果が期待できる施策を優先します。特に、同じ期間でより高いROIが見込める施策は、初期段階で大きなリターンを生み出す可能性が高いです。
    *   **例**: 「低パフォーマンスキャンペーン停止・予算再配分」（900%）は、最も優先度の高い施策の一つと考えられます。

2.  **実施期間（短期 vs 長期）**:
    *   **短期的なインパクト**: すぐに効果が見込める施策（例: 数週間〜数ヶ月で結果が出るもの）は、早期に収益貢献し、次の投資資金を創出する上で重要です。これらは「クイックウィン」として優先すべきです。
    *   **長期的・戦略的なインパクト**: 実施に時間がかかっても、企業の競争優位性や持続的な成長に不可欠な施策（例: ロイヤルティプログラムの刷新、動的価格設定ツールの導入）も、並行して計画的に進める必要があります。
    *   **例**: 月間でのROIが高い施策（例: 購入ファネル改善、割引戦略のパーソナライズ）は早期着手に向きます。

3.  **リスクと実現可能性**:
    *   **低リスク・高確度**: 既存のプロセスやツールで実現可能で、失敗のリスクが低い施策は優先的に実施し、成功体験を積むことが重要です。
    *   **高リスク・高リターン**: 新しい技術導入や大規模な組織変更を伴う施策は、十分な検討と準備期間を要するため、慎重なアプローチが必要です。

4.  **KGI（収益最大化またはコスト最適化）への貢献度**:
    *   直接的にKGIに大きく貢献する施策（例: 売上を大幅に増加させる、コストを劇的に削減する）は、優先度が高くなります。
    *   間接的にKGIに貢献する施策（例: 顧客満足度向上、ブランド認知度向上）も重要ですが、直接的な貢献度とバランスを考慮します。

5.  **相乗効果**: 複数の施策を組み合わせることで、単独で実施するよりも大きな効果が期待できる場合、それらをセットで優先的に実施することを検討します。例えば、ウェブサイトのUX改善とパーソナライズされた製品推奨は相乗効果を生み出す可能性があります。

これらの基準を総合的に判断し、ビジネスの現状と戦略的な方向性に合致する施策から順に実行に移していくことで、最も効果的なKGI達成を目指します。

## Calculate Additional KPIs

### Subtask:
データから、平均注文額 (AOV) や簡略化した顧客ライフタイムバリュー (LTV)、および割引率と売上収益を考慮した潜在的な利益率の代理変数など、追加の主要業績評価指標 (KPI) を算出します。また、販売チャネルごとのコンバージョン率も詳細に計算します。


**Reasoning**:
I will calculate the Average Order Value (AOV), a simplified Customer Lifetime Value (LTV), and a proxy for potential profit margin as per the instructions. This involves basic arithmetic operations on existing columns and will add new columns to the DataFrame.



````python
import numpy as np

# 1. 平均注文額 (AOV) を計算
# 仮に 'num_previous_purchases' を総購入回数の代理変数として使用します。
# より正確には、購入ごとのデータを合計し、注文数で割る必要がありますが、
# 現在のデータ構造では個々の注文がないため、総売上を総購入回数の合計で割ることで近似します。
# または、各レコードが1つの注文と仮定し、sales_revenue_usdの平均と解釈することもできます。
# ここでは、各レコードを1つの注文とし、"num_previous_purchases"を顧客ごとの過去の購入頻度と解釈します。
# したがって、各レコードのsales_revenue_usdが1つの注文に対応すると仮定し、その平均をAOVとします。
# しかし、KPIの定義に基づくと、"総売上収益を総購入回数で割る"とあるため、
# ここでは、簡略化のため、各個別の販売レコードの sales_revenue_usd をAOVの代理変数とします。
# もし、'num_previous_purchases' が各顧客の全期間の購入回数を意味するなら、
# 個々の購入（レコード）には対応しません。
# 指示に従い、"総売上収益を総購入回数で割る"を解釈すると、
# ここでの「総購入回数」がデータフレームのレコード数全体を指す場合も考えられますが、
# より実用的なAOVは個々のトランザクションの平均単価です。
# 既存のデータセットでは、各行が個別のトランザクションと見なせるため、
# sales_revenue_usdをそのままAOVの代理とします。
# あるいは、各行が「顧客の最終購入」であり、"num_previous_purchases"は生涯購入回数と仮定し、
# "sales_revenue_usd"は今回の購入額と解釈します。
# 指示「総売上収益を総購入回数で割る」に基づき、df['sales_revenue_usd']の平均をAOVとします。
# これはデータセットの各行が独立した購入を表すという前提に基づきます。

df['AOV'] = df['sales_revenue_usd']
print("Calculated Average Order Value (AOV).")

# 2. 簡略化した顧客ライフタイムバリュー (LTV) を計算
# AOVに購入頻度（平均購入回数）を掛けることで近似します。
# ここでは、'num_previous_purchases' を購入頻度の代理変数として使用します。
# LTV = AOV * num_previous_purchases
df['simplified_LTV'] = df['AOV'] * (df['num_previous_purchases'] + 1) # +1は今回の購入を含むため
print("Calculated simplified Customer Lifetime Value (LTV).")

# 3. 割引率と売上収益を考慮した潜在的な利益率の代理変数を作成
# これは、sales_revenue_usd に (1 - discount_percentage / 100) を掛けることで求められます。
df['potential_profit_proxy'] = df['sales_revenue_usd'] * (1 - df['discount_percentage'] / 100)
print("Calculated potential profit proxy.")

# 計算したKPIの記述統計量を表示
print("\nDescriptive statistics for new KPIs:")
print(df[['AOV', 'simplified_LTV', 'potential_profit_proxy']].describe())
````

The first step is to identify the top product categories based on total sales revenue, as per the instructions. I will group the DataFrame by 'product_category' and sum the 'sales_revenue_usd' to find the categories with the highest revenue.



````python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calculate total sales revenue for each product category and identify the top categories
total_sales_by_category = df.groupby('product_category')['sales_revenue_usd'].sum().sort_values(ascending=False)
print("Total Sales Revenue by Product Category:")
print(total_sales_by_category)

top_n = 3 # Let's select the top 3 product categories
top_product_categories = total_sales_by_category.head(top_n).index.tolist()
print(f"\nTop {top_n} Product Categories based on total sales revenue: {top_product_categories}")
````

**Reasoning**:
I need to filter the DataFrame for the top product categories and then visualize their sales performance by region and season. I will create two sets of bar plots: one for regional performance and one for seasonal performance, for each of the top product categories.



````python
filtered_df = df[df['product_category'].isin(top_product_categories)]

plt.figure(figsize=(18, 12))

# Regional Performance for Top Product Categories
for i, category in enumerate(top_product_categories):
    plt.subplot(2, len(top_product_categories), i + 1)
    regional_sales = filtered_df[filtered_df['product_category'] == category].groupby('region')['sales_revenue_usd'].mean().sort_values(ascending=False)
    sns.barplot(x=regional_sales.index, y=regional_sales.values, palette='viridis', hue=regional_sales.index, legend=False)
    plt.title(f'{category} Sales by Region')
    plt.xlabel('Region')
    plt.ylabel('Average Sales Revenue (USD)')
    plt.xticks(rotation=45, ha='right')

# Seasonal Performance for Top Product Categories
for i, category in enumerate(top_product_categories):
    plt.subplot(2, len(top_product_categories), len(top_product_categories) + i + 1)
    seasonal_sales = filtered_df[filtered_df['product_category'] == category].groupby('season')['sales_revenue_usd'].mean().sort_values(ascending=False)
    sns.barplot(x=seasonal_sales.index, y=seasonal_sales.values, palette='magma', hue=seasonal_sales.index, legend=False)
    plt.title(f'{category} Sales by Season')
    plt.xlabel('Season')
    plt.ylabel('Average Sales Revenue (USD)')
    plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
````
